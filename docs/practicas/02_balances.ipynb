# ARN y proteína: dos escalas

Integra dos balances, compara estados y revisa unidades. Alfa: u.a. ARN/h; dm: 1/h; beta: u.a. proteína/(u.a. ARN h); dp: 1/h.

Lecciones relacionadas: 00 · Retomar las ecuaciones diferenciales: del balance a la solución, 03 · Expresión génica completa: ARN, proteína, compartimentos y reporteros.

Fuentes de los principios: [A01](https://www.math.uwaterloo.ca/~bingalls/MMSB/), [R05](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html).

## Lectura del código

gen devuelve dos pendientes en el orden ARN, proteína. integrar recibe la función, el estado inicial [0,0] y los tiempos; su resultado coloca estados en filas. y[0] selecciona ARN y y[1] proteína. El primer balance produce plantilla; el segundo la utiliza sin consumirla. El equilibrio [4,60] se calcula poniendo las dos pendientes a cero, no leyendo el último punto. El final de un registro finito solo se aproxima al equilibrio y puede requerir más tiempo.

## Comprueba y cambia
¿Por qué empieza la proteína con pendiente cero? Porque beta*m(0)=0 y p(0)=0. Al duplicar dp, el equilibrio proteico se reduce a 30; el del ARN permanece en 4. No compares unidades de ARN y proteína como si fueran idénticas.
