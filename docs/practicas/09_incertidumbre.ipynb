## Lectura del código

estimaciones empieza como una lista vacía. Cada vuelta genera un conjunto nuevo de datos bajo el modelo supuesto, vuelve a ajustarlo y añade una pareja de parámetros. Convertir la lista en array permite obtener cuantiles por columna; axis=0 resume repeticiones para cada parámetro. El diagrama de dispersión muestra compensación conjunta. Los cuantiles describen este experimento simulado; un bootstrap de datos reales exige un modelo de error y una unidad de remuestreo apropiados.

## Preparación
Ejecuta desde esta carpeta o desde `python/`. Instala las dependencias siguiendo el README.

```python
from pathlib import Path
import sys
base = Path(__file__).resolve().parent.parent if '__file__' in globals() else Path.cwd()
if not (base / 'circuitos.py').exists():
    base = base.parent
sys.path.insert(0, str(base))
import numpy as np
import matplotlib.pyplot as plt
from circuitos import *
```

```python
rng=np.random.default_rng(13)
t=np.linspace(0,20,31)
estimaciones=[]
for i in range(100):
    datos=primer_orden(t,8,.4)+rng.normal(0,1,len(t))
    p,_=ajustar_primer_orden(t,datos)
    estimaciones.append(p)
estimaciones=np.array(estimaciones)
print('Cuantiles 2,5/50/97,5%:',np.quantile(estimaciones,[.025,.5,.975],axis=0))
plt.scatter(*estimaciones.T,s=12)
plt.xlabel('alfa (u.a./h)');plt.ylabel('delta (1/h)');plt.show()
```

## Comprueba y cambia
Solución: los parámetros pueden estar correlacionados aun si el ajuste es bueno. Estos cuantiles describen el procedimiento con verdad conocida; para un bootstrap ajustado a tus datos, simula desde sus parámetros estimados y respeta su modelo de error.
