## Lectura del código

La simulación devuelve tiempos de eventos y estados enteros. La propensión de producción es constante y la de pérdida crece con el número de moléculas. Cada salto añade o retira una unidad; el intervalo se muestrea de una exponencial con la tasa total actual. Una realización no es una media. El dibujo escalonado mantiene el estado entre eventos. Para construir un histograma en tiempo regular debes ponderar duraciones o consultar cada realización en el mismo instante.
