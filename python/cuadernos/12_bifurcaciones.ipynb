# Ramas estables e inestables de una autoactivación

Construye el diagrama de equilibrios de x′=a x²/(1+x²)−x. La variable y el tiempo son adimensionales. Un barrido de trayectorias no muestra por sí solo las ramas inestables.

Lección 07: Bifurcaciones y continuación: cuándo cambia la conducta del circuito.

Fuentes de los principios: [A01](https://www.math.uwaterloo.ca/~bingalls/MMSB/), [A02](https://www.mat.univie.ac.at/~gerald/ftp/book-ode/).

## Lectura del código

La malla de a recorre modelos con distinta capacidad de autoactivación. Las raíces se calculan para cada valor; no se recorre el tiempo. La estabilidad se clasifica con la derivada del campo en cada raíz. Una línea discontinua marca una rama inestable que normalmente no aparece al integrar trayectorias. El punto a=2 se obtiene del discriminante del polinomio; no se estima por el píxel donde dos curvas parecen tocarse. Mantén también la rama x=0.

```python
from pathlib import Path
import sys
base = Path(__file__).resolve().parent.parent if '__file__' in globals() else Path.cwd()
if not (base / 'circuitos_avanzados.py').exists():
    base = base.parent
sys.path.insert(0, str(base))
import numpy as np
import matplotlib.pyplot as plt
import circuitos_avanzados as av
```

```python
a=np.linspace(2,5,301)
low=(a-np.sqrt(a*a-4))/2;high=(a+np.sqrt(a*a-4))/2
plt.plot(a,high,label='Rama alta estable')
plt.plot(a,low,'--',label='Rama intermedia inestable')
plt.plot([0,5],[0,0],label='Origen estable')
plt.xlabel('a (adimensional)');plt.ylabel('Equilibrio x*');plt.legend();plt.show()
print('Raíces para a=2,5:',av.ramas_autoactivacion(2.5))
t=np.linspace(0,30,301)
for x0 in [.4,.6]:
    y=av.integrar(av.autoactivacion,[x0],t,args=(2.5,))
    plt.plot(t,y[0],label=f'x0={x0}')
plt.xlabel('Tiempo adimensional');plt.ylabel('x');plt.legend();plt.show()
```
