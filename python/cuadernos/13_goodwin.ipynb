# Derivar y comprobar el umbral de Goodwin

Compara relajación y oscilación en tres etapas con pérdidas iguales. El umbral analítico es un candidato de Hopf; la estabilidad de los ciclos necesita comprobarse. Todos los valores son adimensionales.

## Lectura del código

El estado contiene ARN x y dos etapas posteriores y,z. La ganancia g procede de la derivada de la producción respecto a z en equilibrio. El programa encuentra s resolviendo la ecuación estacionaria y utiliza ese s para calcular el Jacobiano. g=8 corresponde a pérdidas iguales en este modelo. Las curvas temporales contrastan la predicción local, pero el nacimiento y estabilidad de un ciclo requieren más que localizar una parte real cero.
